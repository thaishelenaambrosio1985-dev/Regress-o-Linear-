```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("sakshisatre/ice-cream-sales-dataset")

print("Path to dataset files:", path)
```

```python
import os
import pandas as pd

# Listar os arquivos no diretório do dataset para encontrar o arquivo CSV
files = os.listdir(path)
print("Arquivos encontrados:", files)

# Encontrar o primeiro arquivo CSV e carregá-lo
csv_files = [f for f in files if f.endswith('.csv')]
if csv_files:
    csv_path = os.path.join(path, csv_files[0])
    df = pd.read_csv(csv_path)
    print(f"DataFrame carregado com sucesso a partir de: {csv_files[0]}")
    display(df.head())
else:
    print("Nenhum arquivo CSV encontrado no diretório.")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Configurar o estilo do gráfico
sns.set_theme(style="whitegrid")

# Criar o gráfico de dispersão com linha de regressão
plt.figure(figsize=(10, 6))
sns.regplot(data=df, x='Temperature', y='Revenue', scatter_kws={'alpha':0.6}, line_kws={'color':'red'})

# Adicionar títulos e rótulos
plt.title('Relação entre Temperatura e Receita (Vendas de Sorvete)', fontsize=14)
plt.xlabel('Temperatura (°C)', fontsize=12)
plt.ylabel('Receita ($)', fontsize=12)

# Mostrar o gráfico
plt.show()
```

```python
from sklearn.linear_model import LinearRegression

# Preparar os dados para o modelo
X = df[['Temperature']]
y = df['Revenue']

# Inicializar e treinar o modelo de regressão linear
model = LinearRegression()
model.fit(X, y)

# Obter o coeficiente angular (slope/inclinação) e o intercepto
slope = model.coef_[0]
intercept = model.intercept_

print(f"Coeficiente de Inclinação (Slope): {slope:.4f}")
print(f"Intercepto (Bias): {intercept:.4f}")
print(f"\nEquação da reta: Receita = {slope:.4f} * Temperatura + ({intercept:.4f})")
```
